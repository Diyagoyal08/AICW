inheritance


```python
class Father:
    def __init__(self , name , age):
        self.name = name
        self.age = age

    def greet(self , greeting):
        print(f"hello  {greeting}, my name is {self.name} and I am {self.age} years old.")

class Son(Father):
    def __init__(self , name , age , school):
        super().__init__(name , age)
        self.school = school

    def greet(self , greeting):
        print(f"hello  {greeting}, my name is {self.name} and I am {self.age} years old. I study in {self.school}.")        
```

```python
Father1 = Father("John", 50)
Father1.greet("hi")

Son1 = Son("Mike", 20 , "ABC School")
Son1.greet("hello")
```

encapsulation 
  getter and setter methods for name and age

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age
```

```python
diya = Person("Diya", 20)
diya.greet()

diya.set_name("Diya goyal")
diya.set_age(21)
diya.greet()

diya.get_name()
diya.get_age()
```
